```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
data = pd.read_csv("heart.csv")

print("Dataset Loaded Successfully")

print(data.head())
```

```python
print(data.shape)

print(data.info())

print(data.isnull().sum())
```

```python
X = data.drop("target", axis=1)

y = data["target"]

print(X.head())

print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

print(X_train.shape)

print(X_test.shape)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)

print("Feature Scaling Completed")
```

```python
dt_model = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)
```

```python
dt_model.fit(X_train, y_train)

print("Decision Tree Model Trained")
```

```python
dt_prediction = dt_model.predict(X_test)

print(dt_prediction)
```

```python
dt_accuracy = accuracy_score(y_test, dt_prediction)

print("Decision Tree Accuracy :",
      round(dt_accuracy * 100, 2), "%")
```

```python
print(confusion_matrix(y_test, dt_prediction))
```

```python
dt_cm = confusion_matrix(y_test, dt_prediction)

print(dt_cm)

plt.imshow(dt_cm)

plt.title("Decision Tree Confusion Matrix")

plt.colorbar()

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```

```python
print(classification_report(y_test, dt_prediction))
```

```python
rf_model = RandomForestClassifier(
    n_estimators=50,
    max_depth=5,
    random_state=42
)
```

```python
rf_model.fit(X_train, y_train)

print("Random Forest Model Trained")
```

```python
rf_prediction = rf_model.predict(X_test)

print(rf_prediction)
```

```python
rf_accuracy = accuracy_score(y_test, rf_prediction)

print("Random Forest Accuracy :",
      round(rf_accuracy * 100, 2), "%")
```

```python
print(confusion_matrix(y_test, rf_prediction))
```

```python
rf_cm = confusion_matrix(y_test, rf_prediction)

print(rf_cm)

plt.imshow(rf_cm)

plt.title("Random Forest Confusion Matrix")

plt.colorbar()

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```

```python
print(classification_report(y_test, rf_prediction))
```

```python
nb_model = GaussianNB()
```

```python
nb_model.fit(X_train, y_train)

print("Naive Bayes Model Trained")
```

```python
nb_prediction = nb_model.predict(X_test)

print(nb_prediction)
```

```python
nb_accuracy = accuracy_score(y_test, nb_prediction)

print("Naive Bayes Accuracy :",
      round(nb_accuracy * 100, 2), "%")
```

```python
print(confusion_matrix(y_test, nb_prediction))
```

```python
nb_cm = confusion_matrix(y_test, nb_prediction)

print(nb_cm)

plt.imshow(nb_cm)

plt.title("Naive Bayes Confusion Matrix")

plt.colorbar()

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```

```python
print(classification_report(y_test, nb_prediction))
```

```python
svm_model = SVC(
    kernel='linear',
    C=0.5
)
```

```python
svm_model.fit(X_train, y_train)

print("SVM Model Trained")
```

```python
svm_prediction = svm_model.predict(X_test)

print(svm_prediction)
```

```python
svm_accuracy = accuracy_score(y_test, svm_prediction)

print("SVM Accuracy :",
      round(svm_accuracy * 100, 2), "%")
```

```python
print(confusion_matrix(y_test, svm_prediction))
```

```python
svm_cm = confusion_matrix(y_test, svm_prediction)

print(svm_cm)

plt.imshow(svm_cm)

plt.title("SVM Confusion Matrix")

plt.colorbar()

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```

```python
print(classification_report(y_test, svm_prediction))
```

```python
models = [
    "Decision Tree",
    "Random Forest",
    "Naive Bayes",
    "SVM"
]

accuracy = [
    dt_accuracy * 100,
    rf_accuracy * 100,
    nb_accuracy * 100,
    svm_accuracy * 100
]

plt.bar(models, accuracy)

plt.xlabel("Algorithms")

plt.ylabel("Accuracy")

plt.title("Accuracy Comparison")

plt.show()
```

```python
print("Decision Tree Accuracy :",
      round(dt_accuracy * 100, 2), "%")

print("Random Forest Accuracy :",
      round(rf_accuracy * 100, 2), "%")

print("Naive Bayes Accuracy :",
      round(nb_accuracy * 100, 2), "%")

print("SVM Accuracy :",
      round(svm_accuracy * 100, 2), "%")
```
